```python
df_clean = df.copy()
```

```python
df_clean.columns = df_clean.columns.str.strip()
```

```python
df_clean["État"] = df_clean["État"].fillna("Inconnu")
```

```python
df_clean["Priorité"] = df_clean["Priorité"].fillna("Inconnu")
```

```python
df_clean["ID de phase"] = df_clean["ID de phase"].fillna("Inconnu")
```

```python
df_clean.isna().sum()
```

```python
print(df_clean["État"].value_counts(dropna=False))
```

```python
print(df_clean["Priorité"].value_counts(dropna=False))
```

```python
print(df_clean["ID de phase"].value_counts(dropna=False))
```

```python
df_clean["Temps_resolution_heures"] = (
    df_clean["Date de résolution"]
    - df_clean["Date/Heure de création"]
).dt.total_seconds() / 3600
```

```python
df_clean[
    [
        "ticket_id",
        "Date/Heure de création",
        "Date de résolution",
        "Temps_resolution_heures"
    ]
].head(20) 
```
