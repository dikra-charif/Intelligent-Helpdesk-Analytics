```python
def predire_sla(priorite, categorie, titre):
    
    ticket = pd.DataFrame([{
        "Priorité": priorite,
        "Categorie": categorie,
        "Titre de la demande": titre
    }])
    
    prediction = model_sla.predict(ticket)[0]
    probabilite = model_sla.predict_proba(ticket)[0, 1]
    
    if prediction == 1:
        resultat = "RISQUE DE DÉPASSEMENT SLA"
    else:
        resultat = "SLA PROBABLEMENT RESPECTÉ"
    
    return resultat, probabilite
```

```python
resultat, probabilite = predire_sla(
    "Moyenne (P3)",
    "Gestion des comptes et mots de passe",
    "Réinitialisation du mot de passe Windows"
)

print(resultat)
print(f"Probabilité : {probabilite:.2%}")
```
